#FILLING THE MISSING VALUES

###LOCATION

```python
data['location'].value_counts()
#we have only 1 null value here
#so can fill anything won't affect
```

```python
data['location']=data['location'].fillna('Sarjapur Road')
```

###SIZE

```python
data['size'].value_counts()
#16 missing values were there so still less thus we are going to replace it with 2 BHK
```

```python
data['size']=data['size'].fillna('2 BHK')
```

###Bath

```python
data['bath']=data['bath'].fillna(data['bath'].median())
#73 missing values
```

```python
data.info()
```

```python
data.isna().sum()
#no null values
```

```python
data['bhk']=data['size'].str.split().str.get(0).astype(int)
#space se split karke first utha liya and then converted from string to int
```

```python
data[data.bhk>20] #problem of outliers
```

```python
data['total_sqft'].unique() #problem of range, there exits some ranges apart from float and int
```

```python
def convertRange(x):
  temp=x.split('-')
  if len(temp)==2:
    return ((float(temp[0])+float(temp[1])/2))/2
  try:
    return float(x)
  except:
    return None

#split the range according to hyphen
#dono ko float mei convert karke then calc mean
#for nan values wr return none if float mei nahi hua convert
#nan values ko handle karne ke liye we have try and except
```

```python
data['total_sqft']=data['total_sqft'].apply(convertRange)
#apply mei kisi function ka reference we need to pass
#har row ka sq feet we will get by this
```

```python
data.head()
```

###PRICE PER SQUARE FEET --> TO REMOVE OUTLIERS

```python
#made new column
data['price_per_sqft']=data['price']*100000/data['total_sqft']
#multiplied as price in lakhs so decimal will be very big otherwise
```

```python
data['price_per_sqft']
```

```python
data['location'].value_counts()
#too much vakues-->so we can't pass this to the model
```

```python
data['location']=data['location'].apply(lambda x: x.strip())
location_count=data['location'].value_counts()

#aage piche se space hat jayega as voh ek naya bana raha tha
```

```python
location_count
```

```python
location_count_less_10=location_count[location_count<=10]
location_count_less_10
```

```python
data['location']=data['location'].apply(lambda x: 'other' if x in location_count_less_10 else x)
#jinka count 10 se kaam hai they will be called others
```
