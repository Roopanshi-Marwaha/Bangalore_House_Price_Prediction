```python
def convertRange(x):
  temp=x.split('-')
  if len(temp)==2:
    return ((float(temp[0])+float(temp[1])/2))/2
  try:
    return float(x)
  except:
    return None

#split the range according to hyphen
#dono ko float mei convert karke then calc mean
#for nan values wr return none if float mei nahi hua convert
#nan values ko handle karne ke liye we have try and except
```

```python
data['total_sqft']=data['total_sqft'].apply(convertRange)
#apply mei kisi function ka reference we need to pass
#har row ka sq feet we will get by this
```

### OUTLIER DETECTION AND REMOVAL

```python
def remove_outliers_sqft(df):
  df_output=pd.DataFrame()
  for key,subdf in df.groupby('location'):
    m=np.mean(subdf.price_per_sqft)

    st=np.std(subdf.price_per_sqft)

    gen_df=subdf[(subdf.price_per_sqft>(m-st)) & (subdf.price_per_sqft<=(m+st))]
    df_output=pd.concat([df_output,gen_df],ignore_index=True)
  return df_output
data=remove_outliers_sqft(data)
data.describe()

#takes a dataframe of data
#then groupby by loaction --> we get key, subdf
#we calculated the mean and standard deviation
#the taking only that price jo mean se ek std idhar udhar ka ho and storing in gen_df
#concatinationg it to df_output
#then describe
#see the max now and notice the difference
```

```python
def bhk_outlier_remover(df):
    exclude_indices = np.array([])
    for location, location_df in df.groupby('location'):
        bhk_stats = {}
        for bhk, bhk_df in location_df.groupby('bhk'):
            bhk_stats[bhk] = {
                'mean': np.mean(bhk_df.price_per_sqft),
                'std': np.std(bhk_df.price_per_sqft),
                'count': bhk_df.shape[0]
            }
        for bhk, bhk_df in location_df.groupby('bhk'):
            stats = bhk_stats.get(bhk-1)
            if stats and stats['count']>5:
                exclude_indices = np.append(exclude_indices, bhk_df[bhk_df.price_per_sqft<(stats['mean'])].index.values)
    return df.drop(exclude_indices,axis='index')

    #making an array storing the indexes which we don't want --> to exclude
    #then groupby by location (loaction,subdataframe)--> for every loc
    #then bhk ke stats stored
    #then har loaction jo bhi ayegi (loc ka subdataframe) usko again grouby by bhk
    #then bhk_stats mei har bhk ke liye 3 cheeze store --> 1. mean-> bhk ka price per sqfeet ka mean 2. the uska std 3. the count karliya voh bhk kitni baar aya hai
    #agar mujhe 3bhk ka flat mila toh what is the lower bound of its price per sq feet? --> a valid lower bound will be isse kam value --> 2 bhk ka mean
    #so agar 3 bhk ka price_per_sq_feet>2 bhk ka mean then ham iss row ko rakh lenge
    #if we have suff enough data (>5) hoga toh we can accurately calc the mean
    #so, that's what we did in our code
    #we used get as value nahi milega toh loop chalega hi nahi otherwise if we simply write it, it will throw some error
```

###FEATURES AND IMPORTING

```python
X=data.drop(columns=['price'])
y=data['price']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,Lasso,Ridge #we are applying 3 models here
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
from sklearn.metrics import r2_score
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
print(X_train.shape)
print(X_test.shape)
```

##APPLYING LINEAR REGRESSION

```python
column_trans = make_column_transformer((OneHotEncoder(sparse_output=False), ['location']),
                                        remainder='passthrough')
#ohe lagaya location column pr as it was the only column which was categorical
```

```python
scaler=StandardScaler()
```

```python
lr = LinearRegression()
```

```python
pipe=make_pipeline(column_trans,scaler,lr)
#so when you do pipe.fit it will go to ohe--> the scale hoga poora data-->linear regression mei fit hoga
```

```python
pipe.fit(X_train,y_train)
```

```python
y_pred=pipe.predict(X_test)
```

```python
r2_score(y_test,y_pred)
```

#APPLYING LASSO

```python
lasso=Lasso()
```

```python
pip=make_pipeline(column_trans,scaler,lasso)
```

```python
pip.fit(X_train,y_train)
```

```python
y_pred_lasso=pipe.predict(X_test)
```

```python
r2_score(y_test,y_pred_lasso)
#R² score mein generally zyada better hota hai (closer to 1)
```
